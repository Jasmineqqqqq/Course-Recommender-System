```python
import json
from pathlib import Path

output_folder = Path("fixed_notebooks")
output_folder.mkdir(exist_ok=True)

for filename in Path(".").glob("*.ipynb"):
    try:
        with open(filename, "r", encoding="utf-8") as file:
            notebook = json.load(file)

        for cell in notebook["cells"]:
            if cell["cell_type"] in ("markdown", "raw"):
                cell.pop("outputs", None)
                cell.pop("execution_count", None)

        with open(output_folder / filename.name, "w", encoding="utf-8") as file:
            json.dump(notebook, file, ensure_ascii=False, indent=2)

        print("Saved:", filename.name)

    except (ValueError, KeyError, TypeError, OSError) as error:
        print("Could not process:", filename.name, "-", error)
```
